```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

Step 1: Create the Data Set

```python
data = {
    "StudyHours": [1,2,3,4,5,6,7,8],
    "Attendence":[40,45,50,60,70,75,80,90],
    "Assignment": [0,0,0,1,1,1,1,1]
}
df=pd.DataFrame(data)

print("="*70)
print("Data Set Created (Logistic Regression)")
print("="*70)

print("\n1.Training Data")
print("-"*70)
print(df)
```

Step 2 : Seperate the Input and Output

```python
X = df[["StudyHours", "Attendence", "Assignment"]]
y = df["Assignment"]

print("\n2. Input Features")
print("-"*70)
print("X =")
print(X)

print("\n Target Variable:")
print("y =", list(y))

print("\nClass 0 = FAIL")
print("Class 1 = PASS")
```

STEP 3 : CREATE LOGISTIC REGRESSION MODEL

```python
model = LogisticRegression()

#Train the model
model.fit(X,y)
```

Step 3 : Get learned Coefficients

```python
b0 = model.intercept_[0]
b1 = model.coef_[0][0]
b2 = model.coef_[0][1]
b3 = model.coef_[0][2]

print("\n3. Learned Coefficients")
print("-"*70)

print("Intercept (b0)=", b0)
print("Study Hours coefficient (b1) =", b1)
print("Attendence coefficient (b2) =", b2)
print("Assignment coefficient (b3) =", b3)
```

Step 5

```python
print("\n4. Logistic Regression Equation")
print("_"*70)

print(
    f"z={b0:.4f}"
    f"+{b1:.4f}x StudyHours"
    f"+{b2:.4f}x Attendence"
    f"+{b3:.4f}x Assignment"
)

print("\nSigmoid Function:")
print("P=1/(1+e^(-z))")
```
